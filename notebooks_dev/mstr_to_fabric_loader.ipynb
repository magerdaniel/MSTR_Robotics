```python
import pandas as pd
import os
from pathlib import Path

# Your lakehouse path
lakehouse_path = "C:\\Users\\danie\\OneLake - Microsoft\\MSTR_db\\dans_lake.Lakehouse\\Tables"

# Get all CSV files from Files section
csv_source_path = "C:\\Users\\danie\\OneLake - Microsoft\\MSTR_db\\dans_lake.Lakehouse\\Files"

csv_files = list(Path(csv_source_path).glob("*.csv"))

for csv_file in csv_files:
    table_name = csv_file.stem  # filename without extension
    
    # Read CSV
    df = pd.read_csv(csv_file, encoding='utf-8')
    
    # Create table directory
    table_path = os.path.join(lakehouse_path, table_name)
    os.makedirs(table_path, exist_ok=True)
    
    # Write as Parquet (Fabric will treat it as Delta)
    parquet_file = os.path.join(table_path, f"{table_name}.parquet")
    df.to_parquet(parquet_file, index=False, engine='pyarrow')
    
    print(f"Converted {csv_file.name} → {table_name} table ({len(df)} rows)")
```

```python
# Preview a specific table
table_to_preview = 'pa_rep_data'
preview_path = f"{lakehouse_path}\\{table_to_preview}\\{table_to_preview}.parquet"

if pd.io.common.file_exists(preview_path):
    df_preview = pd.read_parquet(preview_path)
    print(f"Table: {table_to_preview}")
    print(f"Shape: {df_preview.shape}")
    df_preview.head(10)
else:
    print(f"Table not found: {preview_path}")
```
